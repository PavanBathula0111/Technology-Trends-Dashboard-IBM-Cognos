```python
pd.set_option('display.max_rows', None)
pd.DataFrame(df.dtypes, columns=['DataType'])
```

```python
df=df.drop_duplicates()
```

```python
df['Age_numeric']=df['Age'].replace({'35-44 years old':40, '45-54 years old':50, '25-34 years old':30,
       '55-64 years old':60, '18-24 years old':20, 'Under 18 years old':17,
       '65 years or older':66, 'Prefer not to say':np.nan})
```

```python
df['Age_numeric'].unique()
```

```python
df=df.dropna(subset=['Age_numeric'])
```

```python
tech_cols = {
    'LanguageHaveWorkedWith': 'Language',
    'DatabaseHaveWorkedWith': 'Database',
    'PlatformHaveWorkedWith': 'Platform',
    'WebframeHaveWorkedWith': 'WebFrame',
    'LanguageWantToWorkWith': 'Language',
    'DatabaseWantToWorkWith': 'Database',
    'PlatformWantToWorkWith': 'Platform',
    'WebframeWantToWorkWith': 'WebFrame'
}

for col, name in tech_cols.items():
    counts = (
        df[col]
        .dropna()
        .str.split(';')
        .explode()
        .value_counts()
        .reset_index()
    )

    counts.columns = [name, 'Count']
    counts.to_csv(f'{col}.csv', index=False)

print("Done!")
```
